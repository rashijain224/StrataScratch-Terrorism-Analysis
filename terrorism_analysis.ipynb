```python
incidents_per_year = df.groupby('iyear').size()
incidents_per_year = incidents_per_year.reset_index(name='incidents')
```

```python
plt.figure(figsize=(6,4))
plt.plot(incidents_per_year['iyear'],incidents_per_year['incidents'])
plt.xlabel('Year')
plt.ylabel('Number Of Incidents')
plt.title('Terrorist Incidents Over the Years')
plt.grid(True)
plt.show()
```

```python
incidents_by_region_year = df.groupby(['iyear','region_txt']).size().reset_index(name='incidents')
incidents_by_region_year
```

```python
plt.figure(figsize=(14, 4))
for region in incidents_by_region_year['region_txt'].unique():
    region_data = incidents_by_region_year[
        incidents_by_region_year['region_txt'] == region]
    plt.plot(region_data['iyear'], region_data['incidents'],label=region )

plt.xlabel('Year')
plt.ylabel('Number of Incidents')
plt.title('Terrorist Incidents by Region Over the Years')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.show()
```

```python
df['total_casualties'] = sum(df['nkill'].fillna(0),df['nwound'].fillna(0))
df['total_casualties']
```

```python
yearly_analysis = df.groupby('iyear').agg(incidents=('eventid', 'count'),casualties=('total_casualties', 'sum')).reset_index()
yearly_analysis.head()
```

```python
correlation = yearly_analysis['incidents'].corr(yearly_analysis['casualties'])
print("Correlation : " , correlation)
```

if correlation is close to +1 = incidents and casualties tend to increase together.
around 0 = no linear relationship
close to -1 = when one increases , other tends to decrease.

```python
plt.Figure(figsize = (12,7))
plt.scatter(yearly_analysis['incidents'],yearly_analysis['casualties'])
plt.xlabel('Number of Incidents')
plt.ylabel('Total Casualties')
plt.title('Incidents vs Total Casualties')
plt.grid(True)
plt.show()
```

```python
attack_counts =  df['attacktype1_txt'].value_counts()
attack_counts
```
